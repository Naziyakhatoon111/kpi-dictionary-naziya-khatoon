```
import pandas as pd, numpy as np, random
from datetime import datetime, timedelta

n=10000
np.random.seed(42)
data={
 'order_id':['ORD-'+str(10000+i) for i in range(n)],
 'order_date':[(datetime(2023,1,1)+timedelta(days=random.randint(0,600))).strftime('%m/%d/%Y') for _ in range(n)],
 'city':[random.choice(['Mumbai','Delhi','Pune','Bokaro',None]) for _ in range(n)],
 'qty':[random.choice([1,2,3,5,-1,'two',None]) for _ in range(n)],
 'price':[random.choice([199,499,999,1500,5000,99999]) for _ in range(n)],
 'discount':[random.choice([0,5,10,20,105,None]) for _ in range(n)],
 'order_status':[random.choice(['Paid','paid','PAID','Failed']) for _ in range(n)]
}
df=pd.DataFrame(data)
df=pd.concat([df, df.sample(200)], ignore_index=True)

print("BEFORE Shape:", df.shape)
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())

# CLEANING - NO WARNING VERSION
df['city']=df['city'].fillna('Unknown')
df['discount']=df['discount'].fillna(0)
df=df.drop_duplicates()
df['qty']=df['qty'].replace({'two':2,-1:1})
df['qty']=pd.to_numeric(df['qty'], errors='coerce').fillna(1).astype(int)
df['discount']=df['discount'].replace(105,10)
df['order_date']=pd.to_datetime(df['order_date'], errors='coerce')
df['order_status']=df['order_status'].str.title()
median_price=df[df['price']<10000]['price'].median()
df.loc[df['price']>10000,'price']=median_price

df['month']=df['order_date'].dt.month
df['year']=df['order_date'].dt.year
df['gmv']=df['qty']*df['price']
df['net_revenue']=df['gmv']*(1-df['discount']/100)
df['profit']=df['net_revenue']*0.2

print("\nAFTER Shape:", df.shape)
print(df.isnull().sum())
print(df.head())

df.to_csv('clean_dataset.csv', index=False)
print("\nclean_dataset.csv created!")import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# PART 1: CREATE MESSY RAW DATA
n = 10000
np.random.seed(42)

data = {
    'order_id': ['ORD-'+str(10000+i) for i in range(n)],
    'order_date': [(datetime(2023,1,1) + timedelta(days=random.randint(0,600))).strftime('%m/%d/%Y') for _ in range(n)],
    'city': [random.choice(['Mumbai','Delhi','Pune','Bokaro',None]) for _ in range(n)],
    'qty': [random.choice([1,2,3,5,-1,'two',None]) for _ in range(n)],
    'price': [random.choice([199,499,999,1500,5000,99999]) for _ in range(n)],
    'discount': [random.choice([0,5,10,20,105,None]) for _ in range(n)],
    'order_status': [random.choice(['Paid','paid','PAID','Failed']) for _ in range(n)]
}
df = pd.DataFrame(data)

# add 200 duplicates
dup = df.sample(200)
df = pd.concat([df, dup], ignore_index=True)

print("BEFORE Shape:", df.shape)
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())

# PART 2: CLEANING
df['city'].fillna('Unknown', inplace=True)
df['discount'].fillna(0, inplace=True)
df.drop_duplicates(inplace=True)

df['qty'] = df['qty'].replace({'two':2, -1:1})
df['qty'] = pd.to_numeric(df['qty'], errors='coerce').fillna(1)
df['qty'] = df['qty'].astype(int)

df['discount'] = df['discount'].replace(105, 10)
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['order_status'] = df['order_status'].str.title()

median_price = df[df['price']<10000]['price'].median()
df.loc[df['price']>10000, 'price'] = median_price

# PART 3: FEATURE ENGINEERING
df['month'] = df['order_date'].dt.month
df['year'] = df['order_date'].dt.year
df['gmv'] = df['qty'] * df['price']
df['net_revenue'] = df['gmv'] * (1 - df['discount']/100)
df['profit'] = df['net_revenue'] * 0.2

print("AFTER Shape:", df.shape)
print(df.isnull().sum())
df.head()

# PART 4: EXPORT
df.to_csv('clean_dataset.csv', index=False)
print("clean_dataset.csv created successfully")
```
